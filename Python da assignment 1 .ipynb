```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1 Temperatures:", temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
temperatures_f = (temperatures_w1 * 9 / 5) + 32
print("Temperatures in Fahrenheit:", temperatures_f)

print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
first_three_days = temperatures_w1[:3]

weekend = temperatures_w1[-2:]

middle_three_days = temperatures_w1[2:5]

print("First three days:", first_three_days)
print("Weekend:", weekend)
print("Middle three days:", middle_three_days)
```

```python
temperatures = np.array(
    [
        [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Row 1: Week 1
        [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1],  # Row 2: Week 2
    ]
)

print("2D Array:\n", temperatures)
```

```python
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

print("\nWeek 1 temperatures:", temperatures[0, :])
print("Week 2 temperatures:", temperatures[1, :])

print("\nWeekend temperatures (both weeks):\n", temperatures[:, -2:])
```

```python
import pandas as pd

marks_data = [95, 92, 89, 85, 80]
ranks = ["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]

marks = pd.Series(data=marks_data, index=ranks)
print("Marks Series:\n", marks)
```

```python
print("1st Rank Mark (position 0):", marks.iloc[0])

print("\nTop 3 Ranks (.loc):\n", marks.loc["Rank1":"Rank3"])

print("\n3rd Rank Mark (.iloc):", marks.iloc[2])

print("\nRanks with marks > 90:\n", marks[marks > 90])
```

```python
marks["Rank1"] = 100

marks = marks.drop("Rank5")

print("Updated Series after modification and drop:\n", marks)

cgpa = marks / 10
print("\nCGPA Series:\n", cgpa)
```

```python
import pandas as pd

data = {
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": [
        "Electronics",
        "Clothing",
        "Electronics",
        "Furniture",
        "Clothing",
        "Electronics",
        "Furniture",
        "Clothing",
        "Furniture",
        "Electronics",
    ],
    "Region": [
        "North",
        "South",
        "North",
        "East",
        "West",
        "North",
        "East",
        "West",
        "South",
        "North",
    ],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400],
}

transactions = pd.DataFrame(data)
print(transactions)
```

```python
print("Head (First 5 rows):\n", transactions.head())
print("\nTail (Last 5 rows):\n", transactions.tail())
print("\nShape:", transactions.shape)
print("Column Names:", list(transactions.columns))
print("\nData Types:\n", transactions.dtypes)

print(
    "\nSelected Columns:\n", transactions[["ProductCategory", "Amount"]]
)

print("\nLast 3 Columns:\n", transactions.iloc[:, -3:])

filtered_df = transactions[
    (transactions["Region"] == "North") & (transactions["Amount"] > 200)
]
print("\nFiltered (Region == 'North' & Amount > 200):\n", filtered_df)

print(
    "\nProductCategory Value Counts:\n",
    transactions["ProductCategory"].value_counts(),
)

print("\nUnique Regions:", transactions["Region"].unique())

print(
    "\nMean Amount by Region:\n",
    transactions.groupby("Region")["Amount"].mean(),
)
```

```python
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165

transactions["Discount"] = transactions["Amount"] * 0.10
print("After updating ID 102 and adding Discount:\n", transactions)

transactions = transactions[transactions["TransactionID"] != 109]

transactions = transactions.drop(columns=["Discount"])

print(
    "\nFinal DataFrame (ID 109 removed & Discount deleted):\n", transactions
)
```
