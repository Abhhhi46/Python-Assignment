```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Arun", "Bala", "Charan", "Divya", "Eswar"]

print("Initial age_list:", age_list)
print("Initial name_list:", name_list)
```

```python
name_list.append("Yazhini")
print("\n2a. After append:", name_list)

age_list.insert(2, 30)
print("2b. After insert:", age_list)

name_list.remove("Yazhini")
print("2c. After remove:", name_list)

popped_item = age_list.pop()
print("2d. Popped element:", popped_item, "| age_list:", age_list)

age_list.extend([29, 30, 26])
print("2e. After extend:", age_list)

age_list.sort(reverse=True)
print("2f. Sorted descending:", age_list)

print("2g. Max age:", max(age_list))
print("    Min age:", min(age_list))
print("    Sum of ages:", sum(age_list))

print("\n3a. First element:", name_list[0])
print("3b. Last element:", name_list[-1])
print("3c. Index 2 to 4:", name_list[2:5])
print("3d. Reverse order:", name_list[::-1])
```

```python
student_marks = {
    "Arun": 85,
    "Bala": 90,
    "Charan": 78,
    "Divya": 92,
    "Eswar": 88
}
print("Initial Dictionary:", student_marks)

print("\nb. Mark of Divya:", student_marks["Divya"])

student_marks["Janani"] = 80
print("c. After adding Janani:", student_marks)

student_marks["Arun"] = 82
print("d. After updating Arun:", student_marks)

print("\ne. Keys:", list(student_marks.keys()))
print("   Values:", list(student_marks.values()))
print("   Key-Value Pairs (Items):", list(student_marks.items()))
```

```python
# a. Create set from list
raw_values = ['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i']
my_set = set(raw_values)
print("a. Output of my_set:", my_set)
# Explanation for (a): Sets automatically eliminate duplicate elements and maintain no fixed order. 
# Hence, duplicate 'a' and 'i' are removed.

# b. Attempting assignment
# Explanation for (b): Sets are unordered collections and do not support indexing or item assignment.
# Running `my_set[4] = 's'` throws a TypeError: 'set' object does not support item assignment.
try:
    my_set[4] = 's'
except TypeError as error:
    print("b. Error caught successfully:", error)

# c. Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("\nc. set1:", set1)
print("   set2:", set2)

# d. Union and Intersection
print("d. Union:", set1.union(set2))
print("   Intersection:", set1.intersection(set2))
```
